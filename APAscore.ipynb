```python
# function for taking the center squad and the bottom left square

import pandas as pd

def get_two_square_means(matrix_file, square_size=10):
    # Load matrix
    matrix = pd.read_csv(matrix_file, sep="\t", header=None)
    n = matrix.shape[0]
    
    # --- Center square ---
    center = n // 2
    half = square_size // 2
    start_c = center - half
    end_c = center + half
    center_square = matrix.iloc[start_c:end_c, start_c:end_c]
    mean_center = center_square.mean().mean()
    
    # --- Bottom-left square ---
    start_r = n - square_size
    end_r = n
    start_c = 0
    end_c = square_size
    bottom_left_square = matrix.iloc[start_r:end_r, start_c:end_c]
    mean_bottom_left = bottom_left_square.mean().mean()
    
    # --- Optional combined mean ---
    combined_mean = (mean_center + mean_bottom_left) / 2
    
    return mean_center, mean_bottom_left, combined_mean
```

```python
import os

os.chdir(r"R:\R_STIK\Hi-C\HiC_marta_Oct25\APAplots\scaled")

files = [
    "HiC-REH-EP1_697HF_E2APBX1-HF_peaks_genome.tab",
    "HiC-REH-EP1-Aro_697HF_E2APBX1-HF_peaks_genome.tab",
    "REH_697HF_E2APBX1-HF_peaks_genome.tab"
]

for filename in files:
    mean_center, mean_bottom_left, combined = get_two_square_means(filename, square_size=10)
    print(f"File: {filename}")
    print(f"  Center mean:      {mean_center}")
    print(f"  Bottom-left mean: {mean_bottom_left}")
    print(f"  Combined mean:    {combined}")
    print("-" * 60)
```
